```python
# iterators are advance python concepts that allows you to efficiently iterate through loops and memory management, They provide a way for iterating through collections sequentially without exposing the underlying structure
```

```python
l = [1,2,3,5,6]
iterator = iter(l)
print(type(iterator))
```

```python
iterator # this hides the list items
```

```python
next(iterator) # accessing 1st element
next(iterator) # accessing 2nd element
next(iterator) # accessing 3rd element
```

```python
try:
    next(iterator)
except StopIteration:
    print("max elements reached")
```
