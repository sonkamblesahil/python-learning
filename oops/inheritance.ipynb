```python
# inheritance is a special concept of OOPS that allows a class to inherit the attributes and methods of another class 
# there are different types of inheritances like single inheritance and multiple inheritance and many more

```

```python
# parent class

class Car:
    def __init__(self,windows,doors,engineType):
        self.windows = windows
        self.doors = doors
        self.engineType = engineType
    
    def drive(self):
        print(f"the person will drive the {self.engineType} Car ")
    
    
```

```python
cara = Car(4,5,"petrol")
print(cara.engineType)
print(cara.drive())
```

```python
# tesla inheriting Car class properties (Single inheritance)
class Tesla(Car):
    def __init__(self,windows,doors,engineType,isSelfDriving):
        super().__init__(windows,doors,engineType) # calling the parent class and call the init method from parent class
        self.isSelfDriving = isSelfDriving
    
    def self_driving(self):
        print(f"Tesla supports {self.isSelfDriving}")
    
modelx = Tesla(4,5,"electric",True)
modelx.drive() # using the function of the Parent Class
modelx.self_driving()
modelx.windows
```

```python

class Battery:
    def __init__(self, battery_kwh):
        self.battery_kwh = battery_kwh

    def battery_info(self):
        print(f"Battery capacity: {self.battery_kwh} kWh")


class SmartFeatures:
    def __init__(self, autopilot_version):
        self.autopilot_version = autopilot_version

    def autopilot_info(self):
        print(f"Autopilot version: {self.autopilot_version}")

# multiple inheritance: ModelY inherits from Car + Battery + SmartFeatures
class ModelY(Car, Battery, SmartFeatures):
    def __init__(self, windows, doors, engineType, battery_kwh, autopilot_version):
        Car.__init__(self, windows, doors, engineType)
        Battery.__init__(self, battery_kwh)
        SmartFeatures.__init__(self, autopilot_version)

    def show_model(self):
        print("This is Tesla Model Y")

modely = ModelY(4, 5, "electric", 75, "FSD v12")
modely.show_model()
modely.drive()           # from Car
modely.battery_info()    # from Battery
modely.autopilot_info()  # from SmartFeatures
```
