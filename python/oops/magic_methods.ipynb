```python
# Magic methods (also called dunder methods, short for double underscore) in Python are special methods that start and end with double underscores, like __init__, __str__, __add__.

# They allow you to define how objects of your class behave with built-in operations (like +, len(), print(), etc.)
```

```python
class Person:
    pass

person = Person()
print(dir(person))  # this prints all the magic methods 
```

```python
print(person)
print(person.__str__)
```

```python
# basic magic methods
class Person:
    def __init__(self,name,age):
        self.name= name
        self.age = age
    def __str__(self):  # changes the funtionality of the magic method __str__
        return f"{self.name},{self.age} years old"

person = Person('sahil',22)

print(person)
```
