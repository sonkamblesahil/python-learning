```python
# generator are simpler way to create iterator, they use the yield keyword to produce the values lazily means they dont store values inside the  memory and give them on fly
```

```python
def square(n):
    for i in range(3):
        yield i**2  # if we used normal return statement the code stops but we want more values in loop so generators can be used this will return 0,1,4
    
```

```python
for i in square(10):
    print(i)
```

```python
a = square(4)
a

# using iterator we can get the values
print(next(a))
print(next(a))
```

```python
# generators are used to read large files as they can be used without loading the entire file into the memory
```

```python
def read_file(path):
    with open(path,'r') as file:
        for line in file:
            yield line
```

```python
path = 'larget.txt'
for line in read_file(path):
    print(line)
```
