```python
# abstraction means hiding the complex implementation methods and exposing only the necessary features 
```

```python
from abc import ABC, abstractmethod  # this is a abstract base class its a python module that helps you to create abstract class

class Vehicle(ABC):
    def drive(self):
        print("car is driving")
    
    @abstractmethod
    def start(self):
        pass

class Car(Vehicle):
    def start(self):
        print("Car is started")

def operate_vehicle(vehicle):
    vehicle.start()

car = Car()
operate_vehicle(car)
```
