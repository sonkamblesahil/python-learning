```python
class Dog:
    
    def __init__(self,name,age):  # this is a default constructor
        self.name=name
        self.age=age
    
    def bark(self):
        print(f"{self.name} is woofing")
       

labro = Dog('labro',22)
print(labro.name)
print(labro.age)
labro.bark()

```

```python
#modelling a bank account

class BankAccount:

    def __init__(self,owner,balance):
        self.owner = owner #left side are the object variables and right side are the parmeters
        self.Balance = balance # self refers to the particular object of that class just like this in java
        
    def deposit(self,amount):
        self.Balance+=amount
        print(f"Amount is deposited new Balance is {self.Balance}")
    def withdraw(self,amount):
        if(amount>self.Balance):
            print("Insuffiecint Funds")
        else:
            self.Balance-=amount
            print(f"funds remaining are {self.Balance} and withdrawn amount is {amount}")
    def getBalance(self):
        return self.Balance
    
person1 = BankAccount("sahil",100)
print(person1.Balance)
person1.deposit(1000)
person1.withdraw(10)
person1.getBalance()
    
    
    
```
