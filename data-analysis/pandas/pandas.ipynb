```python
# pandas is a data manipulation library in python, in's widely used for data analysis and data visualization 
# It provides 2 primary data structures 1.series and 2.Dataframe 
# where series is a one dimentional datastructure and dataframe is a 2D datastructure
import pandas as pd
```

```python
# A series in pandas is a one dimensional storage like object that can holf any data type
# by default the indexing is 0 based
data = [1,2,3,4,5]
series =pd.Series(data)
print(series)
print(type(series))
```

```python
# series from dictionary elements
# index values are the keys if converted a dictionary into a series
data = {'a':2,'b':2,'v':3}
series = pd.Series(data)
print('value of a is:',series['a'])
print(series)
print(type(series))
```

```python
# defining our own required index keys 
data = [10,20,30,40]
idx = ['a','b','c','d']
pd.Series(data,idx)
```

```python
## DataFrame is a 2-D structured data in form of rows and columns just like a table in a csv 

# creating a dataframe from dictionary of lists

data = {
    'name':['sahil','krish','john'],

    "age":[22,44,55],

    "cities":["banglore","pune","mumbai"]

}

df = pd.DataFrame(data)

print(df)
```
