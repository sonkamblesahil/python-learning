```python
# polymorphism is a an oops that allows same methods to behave differently
# simply its a way to perform single task in multiple ways this can be achieved with the help od mehtod overiding and interfaces
# method overifing allows a child class to provide a specific implementation of method of parent class in child class

```

```python
class Animal:
    def speak():
        print("animal shouts")

class Dog(Animal):
    def speak(self):
        print("Dog woofs")

class Cat(Animal):
    def speak(self):
        print("Cat meows")

def animal_speak(animal):
    print(animal.speak())

cat = Cat()
cat.speak()

dog = Dog()
dog.speak()
animal_speak(dog)
```

```python
# polymorphism with functions and methods

class shape():
    def area(self):
        return "The area of figure"

class rectangle(shape):
    def __init__(self,width,height):
        self.width= width
        self.height= height
    def area(self):
        return self.width*self.height

rect = rectangle(10,12)
rect.area()

class circle():
    def __init__(self,radius):
        self.radius = radius
    def area(self):
        return 3.14*self.radius*self.radius

def print_area(shape):
    print(f"the area is {shape.area()}")

rect1 = rectangle(1,4)
cir = circle(4)

print_area(rect1)
print_area(cir)

        

       
```
