```python
class Student:
    def __init__(self, name, age, grade):
        self.__name = name
        self.__age = age
        self.__grade = grade

    # Getter methods
    def get_name(self):
        return self.__name

    def get_age(self):
        return self.__age

    def get_grade(self):
        return self.__grade

    # Setter methods
    def set_name(self, name):
        self.__name = name

    def set_age(self, age):
        if age > 0:
            self.__age = age
        else:
            print("Age must be greater than 0")

    def set_grade(self, grade):
        self.__grade = grade

student1 = Student("Rahul", 20, "A")
print(student1.get_name())
print(student1.get_age())
print(student1.get_grade())

student1.set_age(21)
print(student1.get_age())
```
